# **Understanding the given Variables**

**VIN (1-10):** The first 10 characters of the Vehicle Identification Number used to uniquely identify each vehicle.

**County:** The county where the vehicle is registered.

**City:** The city in which the vehicle is located or registered.

**State:** The state where the vehicle is registered.

**Postal Code:** The ZIP/postal code of the vehicle’s registration area.

**Model Year:** The year in which the vehicle model was manufactured.

**Make:** The brand or manufacturer of the vehicle (e.g., Tesla, Nissan).

**Model:** The specific model name of the vehicle.

**Electric Vehicle Type:** The category of EV, such as Battery Electric Vehicle (BEV) or Plug-in Hybrid Electric Vehicle (PHEV).

**Clean Alternative Fuel Vehicle (CAFV) Eligibility:** Indicates whether the vehicle qualifies for clean fuel incentives or programs.

**Electric Range:** The distance the vehicle can travel on electric power alone (usually in miles).

**Base MSRP:** The original base price of the vehicle set by the manufacturer.

**Legislative District:** The government district associated with the vehicle’s registration location.

**DOL Vehicle ID:** A unique identification number assigned by the Department of Licensing.

**Vehicle Location:** The geographic coordinates or general location of the vehicle.

**Electric Utility:** The electricity provider serving the area where the vehicle is registered.

**2020 Census Tract:** A geographic area defined for statistical purposes based on the 2020 census.

# **Exploratory Data Analysis (EDA)**

# **Question 1: What is the total number of electric vehicles?**

```python
print(EV.shape[0])
```

# **Question 2: How has EV adoption changed over the years?**

```python
yearly_growth = EV['Model Year'].value_counts().sort_index()

plt.figure(figsize=(12,6))
yearly_growth.plot(kind='line', marker='o')
plt.title('EV Adoption Trend Over Years')
plt.xlabel('Model Year')
plt.ylabel('Number of EVs')
plt.grid(True)
plt.show()
```

# Insight:
*   EV adoption has increased rapidly after 2020.
*   2023 recorded the highest EV registrations.

*   The dataset shows strong growth in EV popularity.

# **Question 3: Which states have the highest number of EVs?**

```python
state_counts = EV['State'].value_counts().head(10)

plt.figure(figsize=(10,6))
state_counts.plot(kind='bar')
plt.title('Top States by EV Population')
plt.xlabel('State')
plt.ylabel('Number of EVs')
plt.show()
```

# **Insight:**

Washington dominates the dataset because most records belong to Washington state.

# **Question 4: Which cities have the maximum EV population?**

```python
city_counts = EV['City'].value_counts().head(10)

plt.figure(figsize=(12,6))
city_counts.plot(kind='bar',color='lightpink')
plt.title('Top Cities with Highest EV Population')
plt.xlabel('City')
plt.ylabel('Number of EVs')
plt.show()
```

# **Insight:**

Major urban cities have higher EV adoption due to better charging infrastructure and awareness.

# **Question 5: What are the top 10 manufacturers (makes)?**

```python
make_counts = EV['Make'].value_counts().head(10)

plt.figure(figsize=(10,6))
make_counts.plot(kind='area')
plt.title('Top 10 EV Manufacturers')
plt.xlabel('Manufacturer')
plt.ylabel('Number of Vehicles')
plt.show()
```

# **Insight:**

Tesla dominates the EV market with a very large share compared to competitors.

# **Question 6: Which EV models are most commonly used?**

```python
model_counts = EV['Model'].value_counts().head(10)

plt.figure(figsize=(12,6))
model_counts.plot(kind='barh',color='green')
plt.title('Top EV Models')
plt.xlabel('Model')
plt.ylabel('Count')
plt.show()
```

# **Insight:**

Tesla Model Y and Model 3 are the most popular EV models.

# **Question 7: What is the distribution between BEVs and PHEVs?**

```python
ev_type = EV['Electric Vehicle Type'].value_counts()

plt.figure(figsize=(7,7))
ev_type.plot(kind='pie', autopct='%1.1f%%')
plt.title('BEV vs PHEV Distribution')
plt.ylabel('')
plt.show()
```

# **Insight:**

Battery Electric Vehicles (BEVs) are far more common than Plug-in Hybrid Vehicles (PHEVs).

# **Question 8: What is the average electric range?**

```python
avg_range = EV['Electric Range'].mean()
print(avg_range)
```

Average Electric Range = 44.25 miles (71 km)

# **Question 9: How does electric range vary across vehicle makes?**

```python
range_make = EV.groupby('Make')['Electric Range'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))
range_make.plot(kind='pie')
plt.title('Average Electric Range by Make')
plt.xlabel('Make')
plt.ylabel('Average Range')
plt.show()
```

# **Insight:**

Jaguar, Chevrolet, Nissan, and Tesla show higher average electric ranges.

# **Question 10: Is there a relationship between model year and electric range?**

```python
plt.figure(figsize=(10,6))
plt.scatter(EV['Model Year'], EV['Electric Range'])
plt.title('Model Year vs Electric Range')
plt.xlabel('Model Year')
plt.ylabel('Electric Range')
plt.show()
```

```python
print(EV['Model Year'].corr(EV['Electric Range']))
```

# **Insight:**

The negative correlation occurs because many newer entries have zero range values in the dataset.

# **Question 11: Which model year has the highest EV registrations?**

```python
year_registration = EV['Model Year'].value_counts().sort_index()

plt.figure(figsize=(12,6))
year_registration.plot(kind='line',color='green')
plt.title('EV Registrations by Model Year')
plt.xlabel('Model Year')
plt.ylabel('Number of EV Registrations')
plt.xticks(rotation=45)
plt.show()
```

# **Insight:**

The graph clearly shows that EV registrations increased significantly after 2020, with 2023 recording the highest number of EV registrations in the dataset

# **Question 12: Are newer vehicles showing better electric range?**

```python
range_year = EV.groupby('Model Year')['Electric Range'].mean()

plt.figure(figsize=(12,6))
range_year.plot(kind='line', marker='o')
plt.title('Average Electric Range by Model Year')
plt.xlabel('Model Year')
plt.ylabel('Average Electric Range')
plt.grid(True)
plt.show()
```

# **Insight:**

Modern EVs generally provide better battery efficiency and higher range compared to older EVs.

# **Question 13: Distribution based on CAFV Eligibility**

```python
cafv = EV['Clean Alternative Fuel Vehicle (CAFV) Eligibility'].value_counts()

plt.figure(figsize=(10,6))
cafv.plot(kind='bar',color='lightblue')
plt.title('CAFV Eligibility Distribution')
plt.xlabel('CAFV Eligibility')
plt.ylabel('Count')
plt.show()
```

# **Insight:**

 A large number of vehicles have unknown CAFV eligibility because battery research information is unavailable.

# **Question 14:How does the electric range vary among the top EV manufacturers? (Bar Graph)**

```python
# Find top 10 manufacturers
top_makes = EV['Make'].value_counts().head(10).index

# Filter dataset
top_df = EV[EV['Make'].isin(top_makes)]

# Average electric range for each manufacturer
avg_range = top_df.groupby('Make')['Electric Range'].mean().sort_values(ascending=False)

# Plot
plt.figure(figsize=(10,6))
avg_range.plot(kind='barh',color='r')

plt.title("Average Electric Range by Top EV Manufacturers")
plt.ylabel("Manufacturer")
plt.xlabel("Average Electric Range")
plt.xticks(rotation=45)

plt.show()
```

# **Question 15:What is the trend of BEV and PHEV adoption over the years?**

```python
trend_data = EV.groupby(
    ['Model Year', 'Electric Vehicle Type']
).size().unstack()

# Plot
trend_data.plot(figsize=(12,6))

plt.title("BEV vs PHEV Adoption Over Years")
plt.xlabel("Model Year")
plt.ylabel("Number of Vehicles")

plt.grid(True)
plt.show()
```
